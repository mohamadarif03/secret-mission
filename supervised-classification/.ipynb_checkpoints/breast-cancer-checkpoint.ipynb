```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv("../dataset/creditcard.csv")
df.head()
```

```python
df = df.loc[:, ~df.columns.duplicated()]
```

```python
df['Class_Label'] = df['Class'].map({0:'Normal', 1:'Fraud'})
sns.countplot(data=df, x='Class_Label', hue='Class_Label')
```

```python
df.describe()
```

```python
sns.catplot(data=df, x='Class_Label', y='Amount', hue='Class_Label')
```

```python
df = df.drop(columns='Class_Label', axis=1)
corr=df.corr()
sns.heatmap(corr, cmap='RdYlGn', vmin=-1, center=0)
plt.title("Correlation map")
```

```python
x = df.drop(columns='Class', axis=1)
y = df['Class']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix
print(confusion_matrix(y_pred, y_test))
```

```python
from sklearn.metrics import precision_recall_curve
import numpy as np
y_proba = model.predict_proba(x_test)[:, 1]
precision, recall, thresholds = precision_recall_curve(y_test, y_proba)

f1 = 2 * (precision * recall) / (precision + recall)
best_threshold = thresholds[np.argmax(f1)]

best_threshold
```

```python
y_pred = (y_proba > best_threshold).astype(int)
```

```python
RandomForestClassifier(class_weight='balanced')
```

```python
rf = RandomForestClassifier(
    n_estimators=200,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)

rf.fit(x_train, y_train)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)
cm
```
